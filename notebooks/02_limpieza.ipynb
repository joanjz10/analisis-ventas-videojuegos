# 🧹 Notebook 02 — Limpieza de Datos

## ¿Qué haremos aquí?
- Cargar el dataset original desde `data/raw/`
- Corregir el nombre de la columna `Global Sales` (tiene espacio)
- Verificar que todo quedó limpio
- Guardar el dataset limpio en `data/processed/`

```python
import pandas as pd
import os

print("las librerias fueron importadas correctamente")

ruta_raw = "../data/raw/vgsales.csv"

#asignamos la variable df la lectura del archivo csv que se encuentra en la ruta especificada
df = pd.read_csv(ruta_raw)

print("El archivo csv fue cargado correctamente")
print(f"Filas : {df.shape[0]}")
print(f"Columnas : {df.shape[1]}")
```

```python
#mostramos el nombre de las columnas del dataframe

print("Nombres de las columnas: ")
print(df.columns.tolist())
```

```python
#renombramos la columna "Global Sales" a "Global_Sales" para evitar problemas con el espacio en el nombre de la columna
df=df.rename(columns={"Global Sales":"Global_Sales"})

#mostramos el nombre de las columnas del dataframe para verificar que la columna fue renombrada correctamente
print(df.columns.tolist())

```

```python
#verificamos valores nulos en el dataframe
nulos = df.isnull().sum()
print("Valores nulos por columna: ")
print(nulos)
```

```python
#verifivamos los tipos de datos de cada columna del dataframe
print("Tipos de datos por columna: ")
print(df.dtypes)
```

```python
#creamos la carpeta processed dentro de la carpeta data para guardar el archivo csv limpio en que caso de que no este creada
os.makedirs("../data/processed", exist_ok=True)

#creamos la rua para guardar el archivo csv limpio
ruta_processed = "../data/processed/vgsales_limpio.csv"

#guardamos el dataframe limpio en un archivo csv en la ruta especificada sin el indice
# (df.to_csv sirve para escribir un dataframe en un archivo csv)
df.to_csv(ruta_processed, index=False)

print("El archivo csv limpio fue guardado correctamente en la ruta: ", ruta_processed)
```

## 📋 Conclusiones de la Limpieza

- Se cargó el dataset original desde `data/raw/vgsales.csv`
- Se corrigió el nombre de la columna `Global Sales` → `Global_Sales`
- Se verificaron los valores nulos: **ninguno encontrado**
- Se verificaron los tipos de datos: **todos correctos**
- Se guardó el dataset limpio en `data/processed/vgsales_limpio.csv`
