```python
import pandas as pd
import os
import matplotlib.pyplot as plt
import seaborn as sns

print("las librerias fueron importadas correctamente")

#estilo visual de los graficos

#le dice a seaborn que use un estilo visual con fondo blanco y líneas de cuadrícula.
sns.set_style("whitegrid") 

#configura el tamaño de las figuras para que sean más grandes y fáciles de leer. 
plt.rcParams["figure.figsize"] = (12, 6)

#cargamos el dataframe limpio
ruta_limpia = "../data/processed/vgsales_limpio.csv"
df = pd.read_csv(ruta_limpia)

print("Dataset limpio cargado correctamente")
print(f"filas: {df.shape[0]}")
print(f"Columnas: {df.shape[1]}")
```

## 🎯 Pregunta 1 — ¿Cuáles son los géneros más vendidos a nivel global?

```python
# Agrupamos las ventas globales por género y las ordenamos de mayor a menor
ventas_globales_por_genero = df.groupby("Genre")["Global_Sales"].sum().sort_values(ascending=False)

#creamos la grafica

plt.figure(figsize=(10,6))
sns.barplot(
    x = ventas_globales_por_genero.values,
    y= ventas_globales_por_genero.index,
    palette="viridis"
)

#titulos y etiquetas
plt.title ("Generos más vendidos a nivel globlal",fontsize=16)
plt.xlabel("Ventas Globales (millones de unidades)",fontsize=12)
plt.ylabel("Genero",fontsize=12)

#ajusta automáticamente los márgenes para que nada quede cortado.
plt.tight_layout()

#guarda la grafica en la carpeta resultados/graficas con el nombre pregunta_01_generos.png y una resolución de 150 dpi
plt.savefig("../resultados/graficas/pregunta_01_generos.png", dpi = 150)
```

## 🎯 Pregunta 2 — ¿Qué plataforma domina las ventas?

```python
# Agrupamos las ventas globales por plataforma, tomamos el Top 10

ventas_por_plataforma = df.groupby("Platform")["Global_Sales"].sum().sort_values(ascending=False).head(10)

#creamos la grafica

plt.figure(figsize=(10,6))
sns.barplot(
    x=ventas_por_plataforma.values,
    y=ventas_por_plataforma.index,
    palette="magma"
)
#titulos y etiquetas
plt.title("Top 10 plataformas con mayores ventas globales", fontsize=16)
plt.xlabel("Ventas Globlaes Millones de copias",fontsize=12)
plt.ylabel("Plataforma",fontsize=12)

#ajusta automáticamente los márgenes para que nada quede cortado.
plt.tight_layout()

#guarda la grafica en la carpeta resultados/graficas con el nombre pregunta_02_plataformas.png y una resolución de 150 dpi
plt.savefig("../resultados/graficas/pregunta_02_plataformas.png", dpi=150)


```

##🎯 Pregunta 3 — ¿Cómo han evolucionado las ventas por año?

```python
#Agrupamos ventas globales por año

ventas_por_anio = df.groupby("Year")["Global_Sales"].sum().reset_index()

#creamos la grafica

plt.figure(figsize=(14,6))
sns.lineplot(
    data=ventas_por_anio,
    x="Year",
    y="Global_Sales",
    color="Steelblue",
    linewidth=2.5
)

#titulos y etiquetas

plt.title("Evolución de ventas globales por año",fontsize=16)
plt.xlabel("Año",fontsize=12)
plt.ylabel("Ventas globales Millones de copias",fontsize=12)

#ajusta automáticamente los márgenes para que nada quede cortado.
plt.tight_layout()

##guarda la grafica en la carpeta resultados/graficas con el nombre pregunta_03_ventas_por_anio.png y una resolución de 150 dpi
plt.savefig("../resultados/graficas/pregunta_03_ventas_por_anio.png",dpi=150)

```

## 🎯 Pregunta 4 — ¿Qué publishers generan más ingresos?

```python
#agrupamos las ventas globales  por publisher y tomamos el top 10
ventas_por_publisher = df.groupby("Publisher")["Global_Sales"].sum().sort_values(ascending=False).head(10)

#creamos la grafica

plt.figure(figsize=(14,6))
sns.barplot(
    x= ventas_por_publisher.values,
    y=ventas_por_publisher.index,
    palette="coolwarm"
)

#titulos y etiquetas

plt.title("Top de 10 de los publisher con mas ventas globales",fontsize=14)
plt.xlabel("Ventas globales millones de copias",fontsize=12)
plt.ylabel("Publisher",fontsize=12)

##ajusta automáticamente los márgenes para que nada quede cortado
plt.tight_layout()

#guardamos la grafica en la carpeta resultados/graficas con el nombre Pregunta_04_publishers_mas_ingresos
plt.savefig("../resultados/graficas/Pregunta_04_publishers_mas_ingresos.png",dpi=150)
```

## 🎯 Pregunta 5 — ¿Existen diferencias entre mercados? (NA, EU, JP)

```python
#sumamos las ventas totales de cada una de las regiones NA,EU,JP

ventas_na = df["NA_Sales"].sum()
ventas_eu = df["EU_Sales"].sum()
ventas_jp = df["JP_Sales"].sum()

# Creamos una tabla pequeña con los 3 totales

ventas_por_region = pd.DataFrame({
    "Región":["Norteamerica","Europa","Japon"],
    "Ventas":[ventas_na,ventas_eu,ventas_jp]
})

#creamos la grafica

plt.Figure(figsize=(14,6))
sns.barplot(
    data= ventas_por_region,
    x="Región",
    y="Ventas",
    palette= "Set2"
)

#etiquetas y titulos
plt.title("Ventas globales por región",fontsize=14)
plt.xlabel("Region",fontsize=12)
plt.ylabel("Ventas millones de copias",fontsize=12)

#ajusta automáticamente los márgenes para que nada quede cortado
plt.tight_layout()

#guardamos la grafica en la ruta resultados\graficas con el nombre pregunta_05_diferencia_entre_mercados
plt.savefig("../resultados/graficas/pregunta_05_diferencia_entre_mercados.png",dpi=150)

plt.show()
```

## 🎯 Pregunta 6 — ¿Qué juego fue el más vendido de la historia del dataset?

```python
juego_mas_vendido = df.nlargest(10,"Global_Sales")[["Name","Platform","Year","Genre","Global_Sales"]]

#creamos la grafica

plt.figure(figsize=(14,6))
sns.barplot(x="Global_Sales",
            y="Name",
            data=juego_mas_vendido,
            palette="rocket"
            )

#titulos y etiquetas

plt.title("Top 10 de los juegos más vendidos en toda la historia")
plt.xlabel("Ventas globales millones de copias",fontsize=12)
plt.ylabel("Juegos",fontsize=12)

#ajusta automáticamente los márgenes para que nada quede cortado
plt.tight_layout()

#guardamos el grafico en la ruta resultados\graficas con el nombre de pregunta_06_juegos_mas_vendidos
plt.savefig("../resultados/graficas/pregunta_06_juegos_mas_vendidos.png",dpi=150)

plt.show()

juego_mas_vendido
```
