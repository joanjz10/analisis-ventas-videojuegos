```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("Librerias importadas correctamente.")
```

```python
import urllib.request
import os

# PASO 1: Crear la carpeta si no existe (IMPORTANTE)
os.makedirs("../data/raw", exist_ok=True)
print("✓ Carpeta data/raw/ lista para usar")

# PASO 2: Descargar el archivo
url = "https://raw.githubusercontent.com/cajjster/data_files/main/vgsales.csv"
ruta_destino = "../data/raw/vgsales.csv"

# Intentar descargar el archivo. Si hay error, mostrarlo sin bloquear el programa
try:
    urllib.request.urlretrieve(url, ruta_destino)
    print("✓ El archivo se ha descargado correctamente en data/raw/vgsales.csv")
except Exception as e:
    print(f"❌ Error descargando: {e}")
```

```python
df = pd.read_csv(ruta_destino)
print("✓ Archivo cargado en DataFrame 'df' con éxito")

df.head()
```
